```python
import numpy as np
#Q.1 Write a program to create a NumPy 1D-array with 5 elements and perform basic operations like:
#a) Addition of 2 in all the element
arr=np.array([10,23,45,67,89])
addition=arr+2
print(addition)
#b) Multiply 3 with all the elements
product=arr*3
print(product)
#c) Divide every element by 2
division=arr/2
print(division)

```

```python
#Q.2 Questions on Basic NumPy Array:
#a) Reverse the NumPy array: arr = np.array([1, 2, 3, 6, 4, 5])
arr=np.array([1,2,3,6,4,5])
reversed=arr[::-1]
print(reversed)

#b) Find the most frequent value and their indice(s) in the following arrays:
#i. x = np.array([1,2,3,4,5,1,2,1,1,1]) ii. y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3, ])
x = np.array([1,2,3,4,5,1,2,1,1,1])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
valuesx,countx=np.unique(x,return_counts=True)
valuesy,county=np.unique(y,return_counts=True)
print(valuesx)
print(countx)
most_frequent_x=valuesx[np.argmax(countx)]
x_index=np.where(x==most_frequent_x)
print("Most_Frequent_Value in x is: ",most_frequent_x)
print("It's index is: ",x_index)
print(valuesy)
print(county)
max_count_y = np.max(county)
most_frequent_y_values = valuesy[np.where(county == max_count_y)]
print("Most Frequent Value in y are:", most_frequent_y_values)
for val in most_frequent_y_values:
    y_index = np.where(y == val)[0]
    print(f"Indices for value {val}:", y_index)
```

```python
#Q.4 Write program to create an 1-D NumPy array named <<Your Name>> with evenly spaced 25 numbers from 10 to 100 using linspace(). Print the dimensions of the array, shape, total elements, the data type of each element and total number of bytes consumed by the array. Find the transpose of this array using reshape() attribute. Can we do the same with T attribute?
sania=np.linspace(10,100,25)
print(sania)
print(sania.ndim)
print(sania.shape)
print(sania.size)
print(sania.dtype)
print(sania.nbytes)

reshaped_=sania.reshape(-1,1)
print(reshaped_)

transpose_sania=sania[np.newaxis,:].T
print(transpose_sania)
```

```python
#Q5. Create a 2-D Array of three rows and four columns, named ucs420_<your_name>> with following values – 10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35. Compute the mean, median, max, min, unique elements. Reshape the array to four rows and three columns and name it as reshaped_ ucs420_<your_name>>. Resize the array to two rows and three columns and name it as resized_ ucs420_<your_name>>.

ucs420_sania=np.array([[10,20,30,40],[50,60,70,80],[90,15,20,35]])
print(np.mean(ucs420_sania))
print(np.median(ucs420_sania))
print(np.max(ucs420_sania))
print(np.min(ucs420_sania))
values=np.unique(ucs420_sania)
print(values)

reshaped_ucs420_sania=ucs420_sania.reshape(4,3)
print(reshaped_ucs420_sania)
resized_ucs420_sania=np.resize(ucs420_sania,(2,3))
print(resized_ucs420_sania)
```
