```python
#Take your roll number. Extract its individual digits, and multiply each digit by 10 to form your list L (10 numbers in list now).
roll_no=int(input("Enter your Roll No:- "));
list1=[];

while roll_no!=0:
  digit=roll_no%10;
  roll_no=roll_no//10;
  list1.append(digit);
#1 Print L
print(list1);
newlist=list1.copy();
```

```python
#2 Add two numbers of your choice to L — one using append() and one using insert() at a specific position. Print L after each and explain in a comment what changed.
list1.append(int(100));
print(list1);
list1.insert(2,int(20));
print(list1);
#append adds item at the last of the list whereas insert add the element at the desired index as done. 20 is inserted at 2 index.
```

```python
#Remove two elements from L — one using remove() and one using pop(). Print L after each.
list1.remove(20);
print(list1);
list1.pop(3);
print(list1);
```

```python
#Sort L in ascending order using sort(), then descending order using sort(reverse=True). Print both.
list1.sort();
print(list1);
list1.sort(reverse=True);
print(list1);
```

```python
#Use list comprehension to create a new list containing only the elements of L that are greater than the average of L.
print(list1);
total=sum(list1);
avg=total//len(list1);
list2=[x for x in list1 if x>avg];
print(list2);
```
