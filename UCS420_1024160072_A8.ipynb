```python
#Q1.Write a unique paragraph (5-6 sentences) about your favorite topic (e g.sports, technology, food, books, etc.). 1. Convert text to lowercase and remove punctua on using re.

import re
favourite_food="How do I tell you how much I love food? I love to eat savoury and spicy food especially spicy.Not a sweet person but I do get occasional cravings of cake not dry cake but creamy cake. I love to have mathris, ramen , maggie , chips and home cooked food. As I live in hostel , major missing is homecooked food. As I am from mountains and live in plains, there is no availability of pahadi food here.So thats a major missing I am facing rn."

lowercase=favourite_food.lower();
clean_text=re.sub(r"[^\w\s]","",lowercase)
print(clean_text)

```

```python
#2. Tokenize the text into words and sentences.
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.probability import FreqDist
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('punkt_tab')
nltk.download('wordnet')
from nltk.stem import PorterStemmer, WordNetLemmatizer
```

```python
#3. Split using split() and word_tokenize() and compare how Python split and NLTK’s word_tokenize() differ.
python_split_tokens = clean_text.split()
nltk_tokens = word_tokenize(clean_text)
print(python_split_tokens)
print(nltk_tokens)
```

```python
#4. Remove stopwords (using NLTK's stopwords list).
stop_words = set(stopwords.words('english'))
filtered_tokens = []
for word in nltk_tokens:
    if word.isalnum() and word.lower() not in stop_words:
      filtered_tokens.append(word)


print(filtered_tokens)
```

```python
#5. Display word frequency distribu on (excluding stopwords).
freq_dist = FreqDist(filtered_tokens)
for word, frequency in freq_dist.most_common():
    print(f"'{word}': {frequency}")
```

```python
#Q2. Using the same paragraph from Q1: 1. Extract all words with only alphabets using re.findall() 2. Remove stop words using NLTK’s stopword list 3. Perform stemming with PorterStemmer 4. Perform lemma za on with WordNetLemma zer 5. Compare the stemmed and lemma zed outputs and explain when you’d prefer one over the other.

all_words = re.findall(r"[a-zA-Z]+",lowercase)
print("1. Words extracted (first 10):", all_words[:10])


stop_words = set(stopwords.words('english'))
filtered_words = [word for word in all_words if word not in stop_words]
print("\nStep 2 (First 10 words without stopwords):", filtered_words[:10])

stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in filtered_words]
print("\nStep 3 (First 10 stemmed words):", stemmed_words[:10])


lemmatizer = WordNetLemmatizer()
lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_words]
print("\nStep 4 (First 10 lemmatized words):", lemmatized_words[:10])


```

Stemming removes prefixes or suffixes to obtain a root-like form. It may produce words that are not meaningful dictionary words.



Lemmatization converts words into their meaningful dictionary base form.



Stemming is faster, whereas lemmatization is more linguistically accurate.



I would prefer lemmatization when meaningful words are important and stemming when faster text processing is required.

```python
#Q3. Choose 3 short texts of your own (e.g., different news headlines, product reviews). 1. Use CountVectorizer to generate the Bag of Words representa on. 2. Use TfidfVectorizer to compute TF-IDF scores. 3. Print and interpret the top 3 keywords from each text using TF-IDF.
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
texts = [
    "I love spicy food and home cooked food from the mountains.",
    "Living in a hostel makes me miss homecooked food and sweet cakes.",
    "There is no availability of good pahadi food in the plains."
]

bow_vectorizer = CountVectorizer()
bow_matrix = bow_vectorizer.fit_transform(texts)

bow_df = pd.DataFrame(bow_matrix.toarray(), columns=bow_vectorizer.get_feature_names_out())
print("--- Step 1: Bag of Words Matrix ---")
print(bow_df.to_string())


tfidf_vectorizer = TfidfVectorizer()
tfidf_matrix = tfidf_vectorizer.fit_transform(texts)


feature_names = tfidf_vectorizer.get_feature_names_out()

print("\n--- Step 3: Top 3 Keywords Per Text (TF-IDF Interpretation) ---")
for i, text in enumerate(texts):
    print(f"\nText {i+1}: '{text}'")


    row_scores = tfidf_matrix[i].toarray()[0]

    word_score_pairs = list(zip(feature_names, row_scores))
    top_3 = sorted(word_score_pairs, key=lambda x: x[1], reverse=True)[:3]

    for rank, (word, score) in enumerate(top_3, 1):
        print(f"   Rank {rank}: '{word}' (Score: {score:.3f})")
```

Interpretation of Top 3 Keywords using TF-IDF:

1. Text 1: The keywords highlight the importance of spicy food and food from the mountains. They represent the preference for traditional and home-cooked food.

2. Text 2: The keywords reflect hostel life and the feeling of missing home-cooked food and sweet cakes. They describe the experience of living away from home.

3. Text 3: The keywords emphasize the unavailability of Pahadi food in the plains. They highlight the difference in food availability between mountainous and plain regions.

```python
#Q4. Write 2 short texts (4–6 lines each) describing two different technologies (e.g., AI vs Blockchain). 1. Preprocess and tokenize both texts. 2. Calculate: a. Jaccard Similarity using sets b. Cosine Similarity using TfidfVectorizer + cosine_similarity() c. Analyze which similarity metric gives be er insights in your case.
import re
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

text_ai = """
Artificial Intelligence is transforming our modern digital world.
It uses neural networks and machine learning to analyze massive datasets.
Systems can learn from patterns to make intelligent predictions and autonomous choices.
AI models power automation, natural language processing, and advanced robotics applications.
"""

text_blockchain = """
Blockchain technology provides a decentralized ledger for securing digital assets.
It chains data blocks together using advanced cryptography to prevent tampering.
The immutable network allows multiple untrusted participants to securely verify global transactions.
Smart contracts automatically execute rules without relying on central authorities.
"""

def preprocess(text):
    words = re.findall(r"[a-zA-Z]+", text.lower())
    return words

tokens_ai = preprocess(text_ai)
tokens_blockchain = preprocess(text_blockchain)

print("--- Step 1: Preprocessed Tokens ---")
print("AI Tokens (First 10):", tokens_ai[:10])
print("Blockchain Tokens (First 10):", tokens_blockchain[:10])

set_ai = set(tokens_ai)
set_blockchain = set(tokens_blockchain)

intersection = set_ai.intersection(set_blockchain)
union = set_ai.union(set_blockchain)
jaccard_sim = len(intersection) / len(union)

print("\n--- Step 2a: Jaccard Similarity ---")
print("Shared words:", intersection)
print(f"Jaccard Similarity Score: {jaccard_sim:.4f}")

vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform([text_ai, text_blockchain])

cosine_sim = cosine_similarity(tfidf_matrix[0:1], tfidf_matrix[1:2])

print("\n--- Step 2b: Cosine Similarity ---")
print(f"Cosine Similarity Score: {cosine_sim[0][0]:.4f}")
```

**Jaccard Similarity:** The Jaccard similarity score is 0.0519, indicating very low similarity between the AI and Blockchain texts. Only a few words, such as "advanced", "it", "to", and "digital", are common between them.

**Cosine Similarity:** The cosine similarity score is 0.0816, which also indicates low similarity. It measures similarity by comparing the TF-IDF vectors of both texts and considers the importance of words.

**Conclusion:** Both metrics indicate that Artificial Intelligence and Blockchain are different technologies with very little textual similarity. However, cosine similarity provides better insights in this case because it considers the importance of individual words rather than simply checking whether words are shared.

```python
# Q5. Write a short review for a product or service. 1. Use TextBlob or VADER to find polarity & subjec vity for each review. 2. Classify reviews into Posi ve / Nega ve / Neutral. 3. Create a word cloud using the wordcloud library for all posi ve reviews.
from textblob import TextBlob
import pandas as pd
import matplotlib.pyplot as plt
from wordcloud import WordCloud

reviews = [
    "These wireless headphones have amazing sound quality. I absolutely love them.",
    "The battery life is excellent and the design is very comfortable.",
    "I am very happy with this product. It works perfectly.",
    "The sound quality is terrible and the battery drains very quickly.",
    "I am disappointed with these headphones. They are uncomfortable.",
    "The product is okay. Nothing special about it.",
    "The headphones are average and work as expected."
]

sentiments = []

for review in reviews:
    blob = TextBlob(review)

    polarity = blob.sentiment.polarity
    subjectivity = blob.sentiment.subjectivity

    if polarity > 0.1:
        sentiment = "Positive"
    elif polarity < -0.1:
        sentiment = "Negative"
    else:
        sentiment = "Neutral"

    sentiments.append({
        "Review": review,
        "Polarity": polarity,
        "Subjectivity": subjectivity,
        "Sentiment": sentiment
    })

df = pd.DataFrame(sentiments)

print(df.to_string(index=False))


positive_reviews = df[df["Sentiment"] == "Positive"]["Review"]

positive_text = " ".join(positive_reviews)

wordcloud = WordCloud(
    width=800,
    height=400,
    background_color="white"
).generate(positive_text)

plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")
plt.title("Word Cloud of Positive Reviews")
plt.show()
```

TextBlob is used to calculate the polarity and subjectivity of each product review.



Reviews are classified into Positive, Negative, and Neutral categories based on their polarity scores.



Positive reviews contain words such as amazing, excellent, happy, and perfectly.



The word cloud visually represents frequently occurring words in positive reviews. Words appearing more frequently are displayed in larger sizes.

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense

# 1. Choose a paragraph (~100 words) about technology/AI
training_text = """
Artificial intelligence is transforming our modern world at an unbelievable speed.
From automation to natural language processing, intelligent machines are reshaping every industry.
Developers write code to train powerful neural networks on massive datasets to make predictions.
These deep learning models learn complex patterns and make decisions without explicit programming.
As technology continues to advance, we will see even smarter systems assisting human creativity.
Embracing this automated future requires understanding how models process information and generate meaningful results.
Every line of code we run brings us closer to a smarter and more connected global society.
"""

# Tokenize the text
tokenizer = Tokenizer()
tokenizer.fit_on_texts([training_text])
total_words = len(tokenizer.word_index) + 1

# 2. Create input sequences
input_sequences = []
for line in training_text.strip().split('\n'):
    token_list = tokenizer.texts_to_sequences([line])[0]
    for i in range(1, len(token_list)):
        n_gram_sequence = token_list[:i+1]
        input_sequences.append(n_gram_sequence)

# Pad sequences
max_sequence_len = max([len(x) for x in input_sequences])
input_sequences = np.array(pad_sequences(input_sequences, maxlen=max_sequence_len, padding='pre'))

# Create predictors and label
xs, labels = input_sequences[:,:-1], input_sequences[:,-1]
ys = tf.keras.utils.to_categorical(labels, num_classes=total_words)

# Build a simple LSTM model
model = Sequential([
    Embedding(total_words, 64, input_length=max_sequence_len-1),
    LSTM(100),
    Dense(total_words, activation='softmax')
])

model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
model.summary()

# Train the model
print("Training the model...")
model.fit(xs, ys, epochs=150, verbose=1)
```

```python
# 3. Generate 2-3 new lines of text starting from a seed word
def generate_text(seed_text, next_words, model, max_sequence_len):
    for _ in range(next_words):
        token_list = tokenizer.texts_to_sequences([seed_text])[0]
        token_list = pad_sequences([token_list], maxlen=max_sequence_len-1, padding='pre')
        predicted = np.argmax(model.predict(token_list, verbose=0), axis=-1)

        output_word = ""
        for word, index in tokenizer.word_index.items():
            if index == predicted:
                output_word = word
                break
        seed_text += " " + output_word
    return seed_text

# Generate text lines
line_1 = generate_text("artificial", 8, model, max_sequence_len)
line_2 = generate_text("technology", 8, model, max_sequence_len)
line_3 = generate_text("deep", 8, model, max_sequence_len)

print("\n--- Generated Text Lines ---")
print(line_1)
print(line_2)
print(line_3)
```
