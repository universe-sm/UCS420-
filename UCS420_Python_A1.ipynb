```python
#6.2 Prime number
def prime(a):
  if a<2:
    return "Not a prime number";
  is_prime=True;
  for i in range(2,int(math.sqrt(a))+1):
    if a%i==0:
      is_prime=False;
      break;
    else:
      is_prime=True;
  if is_prime==True:
    return "A prime Number";
  else:
    return "Not a prime number";

print(prime(10));
print(prime(83));
```

```python
#6.3 Add 1 to n
def AddN(n):
    s = 0
    for i in range(1, n + 1):
        s += i
    return s

print(AddN(10))   # Outputs: 55
print(AddN(20))   # Outputs: 210
print(AddN(50))   # Outputs: 1275
print(AddN(200))
```

```python
# using function add all odd numbers
def odd_sum(a):
  sum=0;
  for i in range(0,a+1):
    if(i%2!=0):
      sum=sum+i;
  return sum;

n =int(input("Enter a number: "));
print("Sum of odd numbers from 1 to n is: ",odd_sum(n));
```

```python
# using function add prime numbers
def prime_sum(a):
  is_prime=True;
  sum=0
  for i in range(2,a+1):
    for j in range(2,int(math.sqrt(i)+1)):
      if(i%j==0):
        is_prime=False;
        break;
      else:
        is_prime=True;
    if is_prime==True:
      sum=sum+i;
  return sum;

print("Sum of prime numbers from 1 to 19 is: ",prime_sum(19));
print("Sum of prime numbers from 1 to 100 is: ",prime_sum(100));
print("Sum of prime numbers from 1 to 200 is: ",prime_sum(200));
print("Sum of prime numbers from 1 to 500 is: ",prime_sum(500));
```
