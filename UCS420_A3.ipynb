```python
#Q.6 Create a sample dataset (employees.csv) containing information about employees in a company.
employees=pd.read_csv("employees.csv");
#Shape (number of rows and columns) of the DataFrame.
print(employees.shape);
#Summary of the DataFrame that includes the data types and non-null counts for each column.
print(employees.info());
#Generate descriptive statistics.
print(employees.describe());
#Display the first 5 rows and last 3 rows of the dataset
print("First five rows: ",employees.head());
print("Last 3 rows: ", employees.tail(3));
#Calculate the following statistics from the datas
stats=employees.describe();
#The average salary of employees.
print("Average Salary: ", employees["Salary"].mean())
#The total bonus paid to all employees.
print("Total Bonus: ", employees["Bonus"].sum())
#The youngest employee's age.
print("Youngest Employee's Age: ", employees["Age"].min())
#The highest performance rating.
print("Highest Rating: ", employees["Rating"].max())
# Sort the DataFrame by the Salary column in descending order.
employees = employees.sort_values(by="Salary", ascending=False)
print(employees)
#Add a new column that categorizes employees based on their performance rating:
#Categorize based on performance rating
def categorize(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"

employees["Performance_Category"] = employees["Rating"].apply(categorize)
print(employees)

# Identify missing values in the DataFrame.
print("Missing values:\n", employees.isnull().sum())

#Rename the Employee_ID column to ID.
employees = employees.rename(columns={"Employee_ID": "ID"})
print(employees.columns)

#Find all employees who:
#Have more than 5 years of experience.
experienced = employees[employees["Years_of_Experience"] > 5]
print("More than 5 years experience:\n", experienced)

#Belong to the IT department.
it_dept = employees[employees["Department"] == "IT"]
print("IT Department:\n", it_dept)

# Modify the dataset by adding a new column, Tax, which deducts 10% of the Salary.
employees["Tax"] = employees["Salary"] * 0.10
print(employees)

#Save to new CSV
employees.to_csv("modified_employees.csv", index=False)
```
